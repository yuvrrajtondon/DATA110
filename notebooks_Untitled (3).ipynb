Problem Statement
The objective of this project is to develop a multi-class classification machine learning model that predicts a student's adaptability level towards online education. The prediction is based on several factors, including demographic information, educational background, access to technology, internet connectivity, financial condition, device used for online learning, and learning environment.
The dataset classifies each student into one of three adaptability levels: Low, Moderate, or High. By analyzing these factors, the machine learning model can identify how effectively students adapt to online learning environments.

Real-World Significance
Online education has become an essential part of modern learning, especially after the COVID-19 pandemic. However, not all students adapt to online learning in the same way. Factors such as internet availability, financial condition, access to digital devices, and learning environment significantly influence students' learning experiences.
Predicting a student's adaptability level enables educational institutions to identify students who may require additional academic or technological support. This helps improve learning outcomes, reduce dropout rates, and create more personalized learning strategies.

It prepares the dataset for model training by cleaning the data, checking for missing values, removing duplicate records, encoding categorical variables, and splitting the dataset into training and testing sets. Encoding Categorical Variables
The dataset contains categorical features such as Gender, Age, Education Level, Institution Type, Internet Type, and Adaptivity Level. These variables were converted into numerical values using Label Encoding because machine learning algorithms require numerical input for model training.

```python
from sklearn.preprocessing import LabelEncoder

encoder = LabelEncoder()

for column in df.columns:
    df[column] = encoder.fit_transform(df[column])

df.head()
```

Feature Scaling
Feature scaling was performed using StandardScaler to standardize the input features. Scaling helps improve the performance and convergence of algorithms such as Logistic Regression by ensuring that all features have a similar range of values.

```python
from sklearn.preprocessing import StandardScaler

X = df.drop("Adaptivity Level", axis=1)
y = df["Adaptivity Level"]

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

Train-Test Split
The dataset was divided into training and testing sets using an 80:20 ratio. The training data was used to build the machine learning models, while the testing data was used to evaluate their performance on unseen data.

```python
from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X_scaled,
    y,
    test_size=0.2,
    random_state=42
)

print("Training Data:", X_train.shape)
print("Testing Data:", X_test.shape)
```

Why the Selected Algorithms are Suitable
Logistic Regression
Logistic Regression is a supervised classification algorithm that predicts the probability of a student belonging to one of the adaptability classes. It is suitable because the project involves multi-class classification and the target variable has three classes: Low, Moderate, and High.
Naive Bayes
Naive Bayes is a probabilistic classification algorithm based on Bayes' Theorem. It performs well on categorical datasets and is computationally efficient. It provides a good baseline model for comparison with Logistic Regression.

Model Training Process
The dataset was first preprocessed by handling duplicate records, encoding categorical variables using Label Encoding, and scaling the input features using StandardScaler.
The data was then divided into training (80%) and testing (20%) sets. Both Logistic Regression and Naive Bayes models were trained using the training dataset to learn the relationship between the input features and the target variable.

Prediction Process
After training, the models were tested using the unseen testing dataset. The trained models predicted the Adaptivity Level (Low, Moderate, or High) for each student. The predictions were then evaluated using Accuracy, Precision, Recall, F1-Score, and the Confusion Matrix.

```python
##Logistic Regression
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix

log_model = LogisticRegression(max_iter=1000)

log_model.fit(X_train, y_train)

y_pred_log = log_model.predict(X_test)

print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_log))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_log))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_log))
```

```python
##Naive Bayes
from sklearn.naive_bayes import GaussianNB

nb_model = GaussianNB()

nb_model.fit(X_train, y_train)

y_pred_nb = nb_model.predict(X_test)

print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred_nb))
print("\nClassification Report:\n")
print(classification_report(y_test, y_pred_nb))

print("\nConfusion Matrix:\n")
print(confusion_matrix(y_test, y_pred_nb))
```

```python
##Model Comparison
print("Logistic Regression Accuracy:", accuracy_score(y_test, y_pred_log))
print("Naive Bayes Accuracy:", accuracy_score(y_test, y_pred_nb))
```

Model Evaluation
The performance of the machine learning models was evaluated using standard classification metrics. These metrics help determine how well the models predict the Adaptivity Level of students.

The following evaluation metrics were used:
- Accuracy
- Precision
- Recall
- F1-Score
- Confusion Matrix

Accuracy
Accuracy is the percentage of correctly classified observations out of the total observations.
Higher accuracy indicates better overall model performance.

Precision
Precision measures how many of the instances predicted as a particular class are actually correct.
A higher precision value indicates fewer false positive predictions.

Recall
Recall measures how many actual instances of a class are correctly identified by the model.
A higher recall indicates fewer false negative predictions.

F1-Score
The F1-Score is the harmonic mean of Precision and Recall. It provides a balanced measure of model performance, especially when the classes are imbalanced.

Confusion Matrix
The confusion matrix shows the number of correct and incorrect predictions made by the model for each class. It helps identify which classes are classified correctly and where the model makes errors.
