### (a) Create a new column: `AMOUNT_PER_DURATION`


```python
# Create the new column
df1["AMOUNT_PER_DURATION"] = df1["AMOUNT"] / df1["DURATION"]

print("First 5 rows showing the new column:")
display(df1[["AMOUNT", "DURATION", "AMOUNT_PER_DURATION"]].head())
```

# (d) Grouping & Analysis

### (a) Group by `RESPONSE` and find mean `AMOUNT`

### (i) Demonstrate the difference between `view()` and `copy()`


```python
original = np.array([1, 2, 3, 4, 5])

view_array = original.view()
copy_array = original.copy()

# Change the view and copy
view_array[0] = 100
copy_array[1] = 200

print("Original array after changing the view and copy:")
print(original)

print("\nView array:")
print(view_array)

print("\nCopy array:")
print(copy_array)

print("\nExplanation:")
print("- The view shares data with the original array, so changing the view changes the original.")
print("- The copy has independent data, so changing the copy does not change the original.")
```
